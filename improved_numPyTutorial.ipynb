# Fitting Functions with SciPy

Curve fitting is essential for modeling data. We'll use SciPy's `curve_fit` function to fit our data to a model. Let's start by fitting a single Gaussian to our data.


````python
from scipy.optimize import curve_fit

# Define a quadratic function
def quadratic(x, A, B, C):
    return A * x**2 + B * x + C

# Initial guess for the parameters [A, B, C]
initial_guess = [1.0, 1.0, 1.0]  # You can adjust these based on your data

# Perform the curve fit
popt, pcov = curve_fit(quadratic, voltages, currents, p0=initial_guess)

# Extract the parameters
A, B, C = popt

print('Fitted parameters:')
print(f'A (Quadratic Coefficient): {A}')
print(f'B (Linear Coefficient): {B}')
print(f'C (Constant Term): {C}')

# Plot the data and the quadratic fit
plt.figure(figsize=(8, 5))
plt.plot(voltages, currents, 'o', label='Data')
plt.plot(voltages, quadratic(voltages, *popt), 'r-', label='Quadratic Fit')
plt.title('Quadratic Fit to Data')
plt.xlabel('Voltage [V]')
plt.ylabel('Current [A]')
plt.legend()
plt.grid(True)
plt.savefig('Quadratic_Fit.png')
plt.show()
````

## Explanation

1. **Defining the Quadratic Function**: We define a quadratic function \( f(x) = A \times x^2 + B \times x + C \) with coefficients `A`, `B`, and `C`.
   
2. **Initial Guess**: An initial guess for the parameters is provided. Here, we start with `A=1.0`, `B=1.0`, and `C=1.0`. Depending on your data, you might need to adjust these values to achieve a better fit.

3. **Curve Fitting**: Using SciPy's `curve_fit`, we optimize the parameters `A`, `B`, and `C` to fit the quadratic function to the data (`voltages` vs. `currents`).

4. **Extracting Parameters**: The optimized parameters are extracted from `popt` and assigned to `A`, `B`, and `C` for clarity.

5. **Plotting**: We plot the original data points and overlay the fitted quadratic curve to visualize how well the model fits the data.

6. **Saving the Plot**: The plot is saved as `Quadratic_Fit.png` for future reference.

**Note:** If the quadratic fit does not adequately represent your data, consider adjusting the initial guesses or exploring higher-degree polynomials or different fitting models.


## Step 1: Generating Synthetic Data with 3 Gaussians + Constant

In this step, we'll create synthetic data that simulates a radio spectrum composed of three Gaussian components along with a constant offset. This will help us understand how our fitting algorithms work in a controlled environment.

**Components:**

1. **Three Gaussian Functions:**
   - **Gaussian 1:** Amplitude = 50, Mean = -30, Standard Deviation = 5
   - **Gaussian 2:** Amplitude = 80, Mean = 0, Standard Deviation = 10
   - **Gaussian 3:** Amplitude = 60, Mean = 40, Standard Deviation = 7

2. **Constant Offset:** Represents the baseline noise or background signal.

3. **Noise:** Random Gaussian noise added to simulate measurement uncertainties.

**Purpose:**

By generating this synthetic dataset, we can visually inspect the data and later apply fitting techniques to recover the original parameters. This controlled setup ensures that we know the true underlying model, making it easier to validate our fitting results.


````python
import numpy as np
import matplotlib.pyplot as plt

# Seed for reproducibility
np.random.seed(42)

# Generate x values
x_values = np.linspace(-100, 100, 1000)

# Define three Gaussian functions
def gaussian(x, amplitude, mean, stddev):
    return amplitude * np.exp(-((x - mean) ** 2) / (2 * stddev ** 2))

# Parameters for the three Gaussians
params_gauss1 = {'amplitude': 50, 'mean': -30, 'stddev': 5}
params_gauss2 = {'amplitude': 80, 'mean': 0, 'stddev': 10}
params_gauss3 = {'amplitude': 60, 'mean': 40, 'stddev': 7}

# Constant offset
constant = 10

# Generate the synthetic data
y_values = (
    gaussian(x_values, **params_gauss1) +
    gaussian(x_values, **params_gauss2) +
    gaussian(x_values, **params_gauss3) +
    constant +
    np.random.normal(0, 5, x_values.size)  # Adding noise
)

# Store the synthetic data for fitting
synthetic_data = {'x': x_values, 'y': y_values}

print(synthetic_data.keys(), len(synthetic_data["x"]), len(synthetic_data["y"]))

for x, y in zip(synthetic_data["x"][:10],synthetic_data["y"][:10]):
  print("%.2f %.2f"%(x, y))
````

## Step 2: Visualizing the Synthetic Data

Let's visualize the synthetic DATA we generated. The plot shows three distinct Gaussian peaks superimposed on a constant baseline, with added noise to mimic real observational data.

**Key Features:**

- **Gaussian Peaks:** Represent distinct spectral lines or components.
- **Constant Offset:** Indicates the baseline signal or noise level.
- **Noise:** Adds realism to the data, simulating measurement uncertainties.

**Visualization Purpose:**

By inspecting the plot, we can identify the approximate locations and intensities of the Gaussian components, which will guide our initial guesses during the fitting process.


````python
# Plot the synthetic data
plt.figure(figsize=(10, 6))
plt.plot(synthetic_data['x'], synthetic_data['y'], label='Synthetic Data', color='blue')
plt.grid(True)
plt.legend()
plt.savefig('Synthetic_Spectrum.png')
plt.show()
````

## Step 3: Fitting the Synthetic Data with 4 Gaussians + Constant

Now, we'll attempt to fit the synthetic data using a model that includes four Gaussian components plus a constant offset. This step is crucial for understanding model selection and avoiding overfitting.

**Why 4 Gaussians?**

- **Purpose:** Introducing an extra Gaussian allows us to test whether adding more complexity to the model improves the fit significantly.
- **Overfitting Risk:** Fitting more Gaussians than necessary can lead to overfitting, where the model captures noise rather than the underlying signal.

**Fitting Process:**

1. **Model Definition:** We define a `multi_gaussian` function that sums multiple Gaussian functions and adds a constant offset.
2. **Initial Guess:** Providing a reasonable initial guess is essential for the convergence of the fitting algorithm.
3. **Curve Fitting:** Using SciPy's `curve_fit`, we optimize the parameters to fit the model to the data.

**Output:**

The fitted parameters, including the constant offset and the parameters for each Gaussian, are printed for inspection.


````python
from scipy.optimize import curve_fit

# Define a function with multiple Gaussians plus a constant
def multi_gaussian(x, a0, *params):
    """
    Model function: constant + sum of Gaussians
    a0: constant offset
    params: list of parameters for each Gaussian [amp1, mean1, std1, amp2, mean2, std2, ...]
    """
    y = a0
    num_gaussians = len(params) // 3
    for i in range(num_gaussians):
        amplitude = params[3*i]
        mean = params[3*i + 1]
        stddev = params[3*i + 2]
        y += gaussian(x, amplitude, mean, stddev)
    return y

# Initial guess for 4 Gaussians + constant
# [constant, amp1, mean1, std1, amp2, mean2, std2, amp3, mean3, std3, amp4, mean4, std4]
initial_guess_4gauss = [
    10,    # constant
    50, -30, 5,    # Gaussian 1
    80, 0, 10,     # Gaussian 2
    60, 40, 7,     # Gaussian 3
    30, 70, 5      # Gaussian 4 (additional Gaussian for overfitting)
]

# Perform the curve fit with 4 Gaussians
popt_4gauss, pcov_4gauss = curve_fit(
    multi_gaussian,
    synthetic_data['x'],
    synthetic_data['y'],
    p0=initial_guess_4gauss
)

print('Fitted parameters with 4 Gaussians + Constant:')
print(f'Constant Offset: {popt_4gauss[0]}')
for i in range(4):
    amp = popt_4gauss[1 + 3*i]
    mean = popt_4gauss[2 + 3*i]
    stddev = popt_4gauss[3 + 3*i]
    print(f'Gaussian {i+1}: Amplitude={amp}, Mean={mean}, StdDev={stddev}')
````

## Step 4: Visualizing the 4-Gaussian Fit

After fitting the synthetic data with four Gaussians plus a constant, let's visualize how well the model aligns with the data.

**Observations:**

- **Fit Quality:** Assess whether the additional Gaussian significantly improves the fit or if it's merely capturing noise.
- **Overfitting Indicators:** Look for instances where the extra Gaussian might be fitting random fluctuations rather than true spectral components.

**Purpose of Visualization:**

By comparing the synthetic data with the fitted model, we can visually evaluate the effectiveness of adding more Gaussian components and prepare for the statistical tests that will guide our model selection.


````python
# Generate fitted values using the 4-Gaussian model
fitted_4gauss = multi_gaussian(synthetic_data['x'], *popt_4gauss)

# Plot the data and the 4-Gaussian fit
plt.figure(figsize=(10, 6))
plt.plot(synthetic_data['x'], synthetic_data['y'], label='Synthetic Data', color='blue')
plt.plot(synthetic_data['x'], fitted_4gauss, label='4-Gaussian Fit', color='red')
plt.title('Synthetic Spectrum with 4-Gaussian Fit')
plt.xlabel('Velocity Relative to LSR [km/s]')
plt.ylabel('Uncalibrated Antenna Temperature [K]')
plt.grid(True)
plt.legend()
plt.savefig('Synthetic_Spectrum_4Gauss_Fit.png')
plt.show()
````

## Step 7: Performing the Fisher Test (or F-test) for Model Selection

To objectively determine whether adding an extra Gaussian component significantly improves the fit, we'll perform the Fisher test (F-test). This statistical test compares two nested models to assess whether the more complex model provides a better fit to the data.

**Models Compared:**

1. **3-Gaussian + Constant Model**
2. **4-Gaussian + Constant Model**

**Fisher Test Procedure:**

1. **Calculate Sum of Squared Residuals (SSR):** For both models, compute the SSR, which measures the discrepancy between the data and the model.
2. **Determine Degrees of Freedom (DoF):**
   - **3-Gaussian Model:** DoF = Number of Data Points - Number of Parameters
   - **4-Gaussian Model:** DoF = Number of Data Points - Number of Parameters
3. **Compute the F-Statistic:** This statistic quantifies the ratio of the improvement in fit to the increase in model complexity.
4. **Calculate the p-value:** Determines the probability that the observed improvement could occur by chance.
5. **Decision Rule:**
   - **p-value < 0.05:** The more complex model (4 Gaussians) provides a significantly better fit.
   - **p-value ≥ 0.05:** The simpler model (3 Gaussians) is sufficient.

**Interpretation:**

Based on the p-value, we'll decide whether the additional Gaussian is justified or if the simpler model adequately describes the data.


````python
from scipy.stats import f

# Number of data points
n = len(synthetic_data['x'])

# Compute the expected average noise, taking into accout std(A-B) = std(A) +std(B) = 2 std(A)
average_noise = root_average_noise_squared/2**0.5

# Calculate residuals for both models
residuals_3gauss = (synthetic_data['y'] - fitted_3gauss)
residuals_4gauss = (synthetic_data['y'] - fitted_4gauss)

# Calculate Sum of Squared Residuals
ss_res_3gauss = np.sum(residuals_3gauss**2)
ss_res_4gauss = np.sum(residuals_4gauss**2)

# Degrees of freedom
# For 3 Gaussians + constant: p1 = 1 + 3*3 = 10 parameters
# For 4 Gaussians + constant: p2 = 1 + 4*3 = 13 parameters
p1 = 10
p2 = 13
df1 = n - p1
df2 = n - p2

# Calculate the F-statistic
F = ((ss_res_3gauss - ss_res_4gauss) / (p2 - p1)) / (ss_res_4gauss / df2)

# Calculate the p-value
p_value = 1 - f.cdf(F, p2 - p1, df2)

print(f'F-statistic: {F}')
print(f'p-value: {p_value}')

if p_value < 0.05:
    print("The model with 4 Gaussians provides a significantly better fit.")
else:
    print("The model with 4 Gaussians does not provide a significantly better fit. The 3-Gaussian model is sufficient.")
````
